```python
import pandas as pd
import ast

# === Загружаем CSV ===
ground_truth = pd.read_csv("truth.csv")           # размеченные web_id
predictions = pd.read_csv("predictions_rag.csv")  # предсказания RAG

def parse_web_list(s):
    """Парсим строку вида '[1, 2, 3]' в список чисел"""
    return ast.literal_eval(s) if isinstance(s, str) else []

ground_truth["web_list"] = ground_truth["web_list"].apply(parse_web_list)
predictions["web_list"] = predictions["web_list"].apply(parse_web_list)

#  Подсчёт Hit@K
K = 5
hits = []

for _, row in ground_truth.iterrows():
    qid = row["q_id"]
    true_ids = set(row["web_list"])

    # Проверяем, есть ли этот q_id в предсказаниях
    pred_row = predictions[predictions["q_id"] == qid]
    if pred_row.empty:
        continue

    pred_ids = set(pred_row.iloc[0]["web_list"][:K])

    # Hit@K = 1, если есть хотя бы одно совпадение
    hit = 1 if len(true_ids & pred_ids) > 0 else 0
    hits.append(hit)

# Средний результат
hit_at_k = sum(hits) / len(hits) if hits else 0
print(f"Hit@{K}: {hit_at_k:.4f}")

# по каждому запросу
for idx, h in enumerate(hits):
    print(f"q_id={ground_truth.iloc[idx]['q_id']}: hit={h}")
```
